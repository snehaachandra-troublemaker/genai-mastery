```python
from sentence_transformers import SentenceTransformer

model_name = 'all-MiniLM-L6-v2'
model = SentenceTransformer(model_name)
print(f"Model '{model_name}' loaded successfully.")
```

## Embedding Model Loading

This section loads the pre-trained Sentence Transformer model `all-MiniLM-L6-v2`, which is used to convert text into numerical vector representations (embeddings).

```python
embedding1 = model.encode('How do I check my invoice status?')
embedding2 = model.encode('What is the status of my invoice?')
embedding3 = model.encode('Tell me about your refund policy')
embedding4 = model.encode('How do I reset my password?')
embedding5 = model.encode('What is the weather like today?')

```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Reshape embeddings for cosine_similarity function if they are 1D arrays
similarity_score = cosine_similarity(embedding1.reshape(1, -1), embedding2.reshape(1, -1))[0][0]
print(f"Cosine similarity between 'How do I check my invoice status?' and 'What is the status of my invoice?': {similarity_score}")
```

```python
import numpy as np

# Combine all embeddings into a single 2D array
all_embeddings = np.vstack([embedding1, embedding2, embedding3, embedding4, embedding5])

print("Shape of the combined embeddings array:", all_embeddings.shape)
print("First two rows of the combined embeddings array:")
print(all_embeddings[:2])
```

## Cosine Similarity Matrix

This section calculates the cosine similarity between all pairs of the generated sentence embeddings. Cosine similarity measures the cosine of the angle between two non-zero vectors, indicating their semantic similarity.

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate the cosine similarity matrix for all embeddings
similarity_matrix = cosine_similarity(all_embeddings)

print("Cosine Similarity Matrix:")
print(similarity_matrix)
```

## PCA Visualization

This section reduces the dimensionality of the high-dimensional sentence embeddings to 2D using Principal Component Analysis (PCA) and visualizes them on a scatter plot. Each point on the plot represents a sentence, with similar sentences appearing closer together.

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Reduce dimensions to 2D using PCA
pca = PCA(n_components=2)
embeddings_2d = pca.fit_transform(all_embeddings)

print("Shape of 2D embeddings:", embeddings_2d.shape)
print("First two rows of 2D embeddings:\n", embeddings_2d[:2])
```

```python
# Original sentences for labeling
sentences = [
    'How do I check my invoice status?',
    'What is the status of my invoice?',
    'Tell me about your refund policy',
    'How do I reset my password?',
    'What is the weather like today?'
]

# Create a scatter plot of the 2D embeddings
plt.figure(figsize=(10, 8))
plt.scatter(embeddings_2d[:, 0], embeddings_2d[:, 1])

# Label each point with its corresponding sentence
for i, sentence in enumerate(sentences):
    plt.annotate(sentence, (embeddings_2d[i, 0], embeddings_2d[i, 1]), textcoords="offset points", xytext=(5, 5), ha='left')

plt.title('2D PCA Visualization of Sentence Embeddings')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```

## Key Observations

**Similarity Scores:**

The cosine similarity matrix clearly shows that sentences with similar meanings have high similarity scores. For instance, the two invoice-related questions ('How do I check my invoice status?' and 'What is the status of my invoice?') have a very high cosine similarity of approximately 0.839. Conversely, sentences with very different meanings, such as 'How do I check my invoice status?' and 'What is the weather like today?', have very low similarity scores (around 0.018).

**PCA Plot:**

The 2D PCA visualization reinforces these observations:
*   The two invoice-related questions are plotted very close to each other, forming a tight cluster, which visually confirms their strong semantic relationship.
*   The other sentences ('Tell me about your refund policy', 'How do I reset my password?', 'What is the weather like today?') are positioned further apart from each other and from the invoice questions, reflecting their distinct semantic content.

This visualization provides an intuitive way to understand the relationships between the sentences, showing how the model groups semantically similar phrases together in the embedding space.
