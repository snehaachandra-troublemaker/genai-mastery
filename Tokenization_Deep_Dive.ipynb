# Tokenization

Tokenization is a fundamental step in Natural Language Processing (NLP) where text is split into smaller units called tokens. These tokens can be words, subwords, or characters, and they serve as the input for many NLP models. The `transformers` library provides powerful `AutoTokenizer` classes that automatically select the appropriate tokenizer for a given pre-trained model. In this notebook, we are using the `gemma-2-2b-it` model's tokenizer to process text.

```python
from transformers import AutoTokenizer

tokeniser = AutoTokenizer.from_pretrained("google/gemma-2-2b-it")
```

```python
sent1 = "The quick brown fox jumps over the lazy dog."
sent2 = "Reimbursement preauthorization documentation."
sent3 = "नमस्ते, आप कैसे हैं?"

sent1_tokenised = tokeniser(sent1)
sent2_tokenised = tokeniser(sent2)
sent3_tokenised = tokeniser(sent3)

sent1_encodeId = tokeniser.encode(sent1)
sent2_encodeId = tokeniser.encode(sent2)
sent3_encodeId = tokeniser.encode(sent3)


print("Tokenized outputs:")
print(sent1_tokenised)
print(sent2_tokenised)
print(sent3_tokenised)

print("\nConverted IDs to Tokens:")
print(tokeniser.convert_ids_to_tokens(sent1_tokenised.input_ids))
print(tokeniser.convert_ids_to_tokens(sent2_tokenised.input_ids))
print(tokeniser.convert_ids_to_tokens(sent3_tokenised.input_ids))

print("\nEncoded IDs:")
print(sent1_encodeId)
print(sent2_encodeId)
print(sent3_encodeId)

print("\nToken Counts:")
print(f"Sentence 1 token count: {len(sent1_tokenised.input_ids)}")
print(f"Sentence 2 token count: {len(sent2_tokenised.input_ids)}")
print(f"Sentence 3 token count: {len(sent3_tokenised.input_ids)}")
```

## Observations

From the tokenization output, several key observations can be made:

*   **Special Tokens**: Each tokenized sequence starts with `<bos>` (Beginning Of Sentence) token, which is common in many transformer models to indicate the start of an input sequence.
*   **Subword Tokenization**: The tokenizer employs subword tokenization, especially evident in words like `quick` (preceded by '▁' indicating it's a continuation of a word) or `imbursement`. This strategy helps handle out-of-vocabulary words and reduces the vocabulary size.
*   **Character-level vs. Subword for Non-English**: For the Hindi sentence ("नमस्ते, आप कैसे हैं?"), the tokenization appears to break down words into smaller units or even character-level components, which can be seen from the individual Hindi characters in the `convert_ids_to_tokens` output. This suggests that the tokenizer might not have full word-level understanding for all languages or relies more heavily on subword units for complex scripts.
*   **Attention Mask**: The `attention_mask` is consistently `[1, 1, ..., 1]` for all tokens, indicating that all tokens are part of the active sequence and none are padding tokens in this specific example.

## Comparison

Comparing the token counts and tokenized outputs across the three sentences reveals important aspects of the tokenizer's behavior:

1.  **Sentence 1 (English, common words)**: "The quick brown fox jumps over the lazy dog." resulted in 11 tokens. This sentence, composed of common English words, is efficiently tokenized, with most words getting a single token (possibly prefixed with '▁' to denote subword).
2.  **Sentence 2 (English, complex words)**: "Reimbursement preauthorization documentation." resulted in 7 tokens. Despite being shorter in character count than Sentence 1, its complex and longer words (e.g., 'Reimbursement', 'preauthorization', 'documentation') are broken down into subword units, yet it leads to a lower token count overall due to the fewer distinct 'words' after subword segmentation.
3.  **Sentence 3 (Hindi)**: "नमस्ते, आप कैसे हैं?" resulted in 10 tokens. This demonstrates how tokenization handles different languages. The tokenizer breaks down the Hindi characters into distinct subword units, leading to a token count comparable to the first English sentence, even though the number of discernible 'words' might be fewer in the original sentence. This highlights the tokenizer's ability to represent non-Latin script, often through more granular subword or character-level representations.
